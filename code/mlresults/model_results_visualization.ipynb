```python
# IMPORTS and DATA LOADS
import altair as alt
import pandas as pd

df = pd.read_csv("results.csv")

# Accuracy rows filter
accuracy_df = df[df.metric == "Accuracy"]

# Define model_order for consistent color scaling
model_order = ['Decision Tree', 'Random Forest', 'Logistic Regression']
```

```python
# BAR CHART
import altair as alt
import pandas as pd

x_domain = max(accuracy_df["value"]) + 0.2
bar = alt.Chart(accuracy_df).mark_bar(
).encode(
    x=alt.X(
        "value:Q",
        title="Accuracy",
        scale=alt.Scale(domain=[0, x_domain]),
        axis=alt.Axis(titleFont='Times New Roman', labelFont='Times New Roman')
    ),
    y=alt.Y(
        "model:N",
        sort="-x",
        title=None,
        axis=alt.Axis(labelFont='Times New Roman')
    ),
    color=alt.Color(
        "model:N",
        scale=alt.Scale(
            domain=model_order,
            range=["#a2d7ff", "#fdaaaa", "#b6d7a8"] # blue, red, orange
        ),
        legend=alt.Legend(
                          orient="none",
                          legendX=0,
                          legendY=180,
                          titleFontSize=11,
                          labelFontSize=10,
                          columns = 3,
                          title = 'Model',
                          titleFont='Times New Roman',
                          labelFont='Times New Roman'
                      )
    ),
    tooltip=[
        alt.Tooltip("model:N", title="Model"),
        alt.Tooltip("value:Q", title="Accuracy", format=".2%")
    ]
)

labels = bar.mark_text(
    align="left",
    baseline="middle",
    dx=5,
    fontSize=10,
    font='Times New Roman' # Set font for bar labels
).encode(
    text=alt.Text("value:Q", format=".2%"),
    color=alt.value('black') # Explicitly encode color to black
)

bar_chart = (bar + labels).properties(
    title=alt.TitleParams(text="Model Accuracy Comparison", font='Times New Roman'), # Corrected title font setting
    width=400,
    height=100 # Decreased height for less tall bars
)
bar_chart
```

```python
# SCATTERPLOT TIME vs ACCURACY
scatter = alt.Chart(accuracy_df).mark_circle(size=200).encode(
    x=alt.X(
        "training_time:Q",
        title="Training Time (seconds)",
        scale=alt.Scale(zero=False),
        axis=alt.Axis(titleFont='Times New Roman', labelFont='Times New Roman')
    ),
    y=alt.Y(
        "value:Q",
        title="Accuracy",
        scale=alt.Scale(domain=[0.6, 0.8]),
        axis=alt.Axis(titleFont='Times New Roman', labelFont='Times New Roman')
    ),
    color=alt.Color(
        "model:N",
        title="Model",
        scale=alt.Scale(
            domain=model_order,
            range=["#a2d7ff", "#fdaaaa", "#b6d7a8"] # blue, red, orange
        ),
        legend=alt.Legend(titleFont='Times New Roman', labelFont='Times New Roman')
    ),
    tooltip=[
        alt.Tooltip("model:N", title="Model"),
        alt.Tooltip("training_time:Q", title="Training Time (s)", format=".2f"),
        alt.Tooltip("value:Q", title="Accuracy", format=".4f")
    ]
).properties(
    title=alt.TitleParams(text="Accuracy vs Training Time Trade-off", font='Times New Roman'), # Corrected title font setting
    width=350,
    height=250
)

scatter
```

```python
# PARALLELL COORDINATES

# Metric order for clarity
metric_order = ["Accuracy", "Precision", "Recall", "F1"]

base = alt.Chart(df).encode(
    x=alt.X(
        "metric:N",
        sort=metric_order,
        title="Evaluation Metric",
        axis=alt.Axis(labelAngle=0, titleFont='Times New Roman', labelFont='Times New Roman')
    ),
    y=alt.Y(
        "value:Q",
        title="Score",
        scale=alt.Scale(domain=[0.6, 0.8]),
        axis=alt.Axis(titleFont='Times New Roman', labelFont='Times New Roman')
    ),
    color=alt.Color(
        "model:N",
        title="Model",
        scale=alt.Scale(
            domain=model_order,
            range=["#a2d7ff", "#fdaaaa", "#b6d7a8"] # blue, red, orange
        ),
        legend=alt.Legend(titleFont='Times New Roman', labelFont='Times New Roman')
    ),
    detail="model:N"
)

# Lines (parallel coordinates)
lines = base.mark_line(
    strokeWidth=3,
    opacity=0.8
)

# Points on each metric
points = base.mark_circle(
    size=120,
    filled=True
)

# Numeric labels near points
labels = base.mark_text(
    dy=-12,
    fontSize=10,
    font='Times New Roman' # Set font for point labels
).encode(
    text=alt.Text("value:Q", format=".2f"),
    color=alt.value('black')
)

parallel_coordinates = (lines + points + labels).properties(
    title=alt.TitleParams(text="Model Performance Metrics Comparison", font='Times New Roman'), # Corrected title font setting
    width=400,
    height=250
)

parallel_coordinates
```

```python
viz_layout = bar_chart | scatter | parallel_coordinates
viz_layout
```
